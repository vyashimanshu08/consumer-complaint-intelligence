```python
from pathlib import Path
import pandas as pd
```

```python
PROJECT_ROOT = Path.cwd().parent

NLP_DIR = PROJECT_ROOT / "data" / "nlp_processed"

print("NLP directory:")
print(NLP_DIR)
```

```python
expected_files = [
    NLP_DIR / f"complaints_nlp_part_{i:04d}.csv"
    for i in range(1, 87)
]

missing_files = [
    file.name
    for file in expected_files
    if not file.exists()
]

print("Expected files:", len(expected_files))
print("Missing files:", len(missing_files))

if missing_files:
    print("\nMissing files:")
    for file in missing_files:
        print(file)
else:
    print("\n✅ All 86 NLP files are present.")
```

```python
# Validate structure 
```

```python
validation_results = []

for i, file_path in enumerate(expected_files, start=1):

    print(f"Checking chunk {i:04d}...", end=" ")

    try:
        df_chunk = pd.read_csv(
            file_path,
            usecols=[
                "Complaint ID",
                "Consumer complaint narrative",
                "has_narrative",
                "clean_narrative"
            ]
        )

        validation_results.append({
            "chunk": i,
            "rows": len(df_chunk),
            "columns_ok": True,
            "narratives": int(df_chunk["has_narrative"].sum()),
            "clean_non_null": int(df_chunk["clean_narrative"].notna().sum()),
            "clean_empty": int(
                df_chunk["clean_narrative"]
                .fillna("")
                .str.strip()
                .eq("")
                .sum()
            ),
            "redactions_remaining": int(
                df_chunk["clean_narrative"]
                .fillna("")
                .str.contains(
                    r"\bx{2,}\b",
                    regex=True
                )
                .sum()
            )
        })

        print("OK")

    except Exception as e:
        print(f"ERROR: {e}")
```

```python
validation_df = pd.DataFrame(validation_results)

validation_df
```

```python
print("========== NLP VALIDATION SUMMARY ==========")

print("Chunks checked:", len(validation_df))

print(
    "Total rows:",
    f"{validation_df['rows'].sum():,}"
)

print(
    "Total narratives:",
    f"{validation_df['narratives'].sum():,}"
)

print(
    "Total non-null clean narratives:",
    f"{validation_df['clean_non_null'].sum():,}"
)

print(
    "Total empty clean narratives:",
    f"{validation_df['clean_empty'].sum():,}"
)

print(
    "Redactions remaining:",
    f"{validation_df['redactions_remaining'].sum():,}"
)
```

```python
problem_cases = []

for i, file_path in enumerate(expected_files, start=1):

    df_chunk = pd.read_csv(
        file_path,
        usecols=[
            "Complaint ID",
            "Consumer complaint narrative",
            "has_narrative",
            "clean_narrative"
        ]
    )

    problems = df_chunk[
        (df_chunk["has_narrative"] == True) &
        (
            df_chunk["clean_narrative"].isna() |
            df_chunk["clean_narrative"].fillna("").str.strip().eq("")
        )
    ]

    if len(problems) > 0:
        problem_cases.append(problems)

problem_df = pd.concat(problem_cases, ignore_index=True)

print("Problematic narratives:", len(problem_df))
```

```python
problem_df[
    [
        "Complaint ID",
        "Consumer complaint narrative",
        "clean_narrative"
    ]
].head(20)
```

```python
# Identify narratives that contain only CFPB redaction tokens
# after removing spaces and date-style redaction patterns.

def is_redaction_only(text):
    if not isinstance(text, str) or not text.strip():
        return False

    tokens = text.lower().split()

    return all(
        token.startswith("x") and set(token) == {"x"}
        for token in tokens
    )


redaction_only = problem_df[
    problem_df["Consumer complaint narrative"].apply(is_redaction_only)
]

print("Total problematic narratives:", len(problem_df))
print("Redaction-only narratives:", len(redaction_only))
print(
    "Other problematic narratives:",
    len(problem_df) - len(redaction_only)
)
```

```python
other_problematic = problem_df[
    ~problem_df["Complaint ID"].isin(redaction_only["Complaint ID"])
]

print("Other cases:", len(other_problematic))

pd.set_option("display.max_colwidth", 500)

other_problematic[
    ["Complaint ID", "Consumer complaint narrative"]
].head(30)
```

```python
remaining_redaction_cases = []

for file_path in expected_files:

    df_chunk = pd.read_csv(
        file_path,
        usecols=["Complaint ID", "clean_narrative"]
    )

    mask = (
        df_chunk["clean_narrative"]
        .fillna("")
        .str.contains(r"\bx{2,}\b", regex=True)
    )

    if mask.any():
        remaining_redaction_cases.append(
            df_chunk.loc[mask]
        )

remaining_redaction_df = pd.concat(
    remaining_redaction_cases,
    ignore_index=True
)

remaining_redaction_df
```

```python
 for _, row in remaining_redaction_df.iterrows():
    print("=" * 80)
    print("Complaint ID:", row["Complaint ID"])
    print("Clean narrative:")
    print(row["clean_narrative"])
```
